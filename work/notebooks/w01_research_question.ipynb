## 2. The question: decision, action, cost of a wrong call

**Research question:** Among pages with enough search exposure to evaluate, which pages should a content reviewer prioritize for refresh review based on observable evidence of declining performance or opportunity?

**Decision:** Decide which pages should be placed near the top of a limited refresh-review queue.

**Action:** A content or SEO reviewer could inspect a high-ranked page, investigate the reason codes and supporting signals, and decide whether a refresh, further investigation, or no action is appropriate.

**Cost of a wrong call:** A false positive can waste limited editorial or SEO effort on a page that does not need attention. A false negative can cause a genuinely useful review candidate to be missed. The system therefore needs to support human review rather than automatically deciding that a page must be changed.

This is not just "train a model" because the useful output is an interpretable ranked list that supports a real review decision. The model is only one possible component of that decision-support process.
