## 2. Target or proxy

The starter dataset provides `trend_direction`, which can be used as a proxy for whether a page is declining.

For this framing, I define:

- `target_declining = 1` when `trend_direction == "down"`
- `target_declining = 0` otherwise

This is a proxy rather than a direct measure of whether a content refresh would succeed. A declining page is a potential review candidate, but the label does not prove that refreshing the page will cause recovery.

The target is therefore useful for testing whether a ranking model can identify pages associated with declining performance, while keeping the final action with a human reviewer.

## 3. Success metric

**Primary metric: Precision@50**

Precision@50 measures the proportion of the 50 highest-ranked pages that are actually classified as declining under the proxy label.

This metric matches the real action because the reviewer has limited capacity. If the system produces a top-50 review queue, I want as many genuinely declining pages as possible to appear near the top of that queue.

The metric evaluates the ranking quality of the review queue, not whether a refresh itself will improve performance.

## 5. Why ML beats a fixed rule here

A fixed rule could rank pages using one manually chosen condition, such as selecting pages whose trend is down or whose impressions are above a particular threshold.

The limitation is that page performance depends on multiple signals at once. Search exposure, position, content age, content characteristics, and other observable features may interact in ways that are difficult to capture with one fixed threshold.

ML is useful here because it can learn a ranking signal from multiple features and produce a prioritized queue rather than relying on one manually selected cutoff.

The comparison should still be against a simple baseline. ML is only useful if it improves the review queue under the chosen validation setup and remains interpretable enough to support human review.
